```
from pyspark.sql.functions import from_unixtime,unix_timestamp
from pyspark.sql.types import StringType
#converting data format to yyyy-MM-dd HH:mm
def format_date(datecolumn):
    return from_unixtime(unix_timestamp(datecolumn,'MM/dd/yyyy HH:mm'),'yyyy-MM-dd HH:mm')
```

```
spark.conf.set("spark.sql.parquet.int96RebaseModeInWrite", "CORRECTED")
spark.conf.set("spark.sql.legacy.timeParserPolicy", "LEGACY")
```

```
order1=order.withColumn('f_order_purchase_timestamp',format_date(order["order_purchase_timestamp"])) \
.withColumn('f_order_approved_at',format_date(order["order_approved_at"]))\
.withColumn('f_order_delivered_timestamp',format_date(order["order_delivered_timestamp"]))\
.withColumn('f_order_estimated_delivery_date',format_date(order["order_estimated_delivery_date"]))\
.drop('order_purchase_timestamp','order_approved_at','order_delivered_timestamp','order_estimated_delivery_date')

order1.display()
```
